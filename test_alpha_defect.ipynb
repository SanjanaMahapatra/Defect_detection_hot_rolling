```python
import numpy as np
import pandas as pd
import joblib

# Loading the saved model

model = joblib.load('alpha_defect_model.pkl')
scaler = joblib.load('scaler.pkl')
imputer = joblib.load('imputer.pkl')
selected_features = joblib.load('selected_features.pkl')


# Loading the test data

test_df = pd.read_csv('dataset/test.csv')

coil_ids = test_df['CoilID']

X_test = test_df.drop('CoilID', axis=1)

print(test_df.head())
print(test_df.shape)
print(test_df.columns)
```

```python
# Applying preprocessing steps

# Imputing missing values
X_test_imputed = imputer.transform(X_test)

# Scaling the features
X_test_scaled = scaler.transform(X_test_imputed)

feature_names = [f'X{i}' for i in range(1, 50)]
X_test_scaled_df = pd.DataFrame(X_test_scaled, columns=feature_names)

X_test_final = X_test_scaled_df[selected_features]
```

```python
# Making predictions
y_pred = model.predict(X_test_final)
```

```python
# Creating the submission file

submission = pd.DataFrame({
    'CoilID': coil_ids,
    'Y': y_pred
})

submission['Y'] = submission['Y'].astype(int)

submission.to_csv('submission.csv', index=False)

print("Submission file 'submission.csv' created successfully.")
print(f"Shape: {submission.shape}")
print(submission.head())
```
